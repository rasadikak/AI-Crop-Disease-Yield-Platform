```python
import pandas as pd
import numpy as np
```

```python
world_df= pd.read_csv("yield_df.csv")
sl_df= world_df[world_df["Area"]=="Sri Lanka"]
print(sl_df.shape)
print(sl_df.columns)
```

```python
sl_df["kg/ha_yeild"]= sl_df["hg/ha_yield"]/10

print(sl_df.columns)
```

```python
print(sl_df["Item"].unique())
print(sl_df["Year"].min(), sl_df["Year"].max())
print(sl_df.head(10))
```

```python
sl_df= sl_df.drop(columns=["hg/ha_yield"])
```

```python
sl_df= sl_df.rename(columns={"Unnamed: 0":"id"})
sl_df.head(3)
sl_df.shape
```

```python
# How many rows do you have?
print(sl_df.shape)

# What crops are in the dataset?
print(sl_df["Item"].unique())

# What year range?
print(sl_df["Year"].min(), "→", sl_df["Year"].max())

# Any missing values?
print(sl_df.isnull().sum())

# Basic statistics
print(sl_df.describe())
```

```python
# Check rainfall issue
print(sl_df["average_rain_fall_mm_per_year"].nunique())
# If this prints 1 — confirmed useless column

# Also check the id column — probably useless too
print(sl_df["id"].nunique())

# Look at yield range per crop — important
print(sl_df.groupby("Item")["kg/ha_yeild"].agg(["mean", "min", "max"]))
```
